```python
import pandas as pd
import numpy as np
```

```python
# 1. Generate synthetic dataset

np.random.seed(42)  # for reproducibility

students = [f"Student_{i+1}" for i in range(20)]
subjects = ["Math", "Science", "English", "History", "Art"]

# Generate scores between 60 and 100
data = np.random.randint(60, 101, size=(20, 5))

df = pd.DataFrame(data, columns=subjects, index=students)

print(" Initial Dataset:\n")
print(df)
print("\n" + "="*50)

```

```python
# 2. Overall class statistics

print(" Overall Class Statistics:\n")
print("Mean:\n", df.mean())
print("\nMax:\n", df.max())
print("\nMin:\n", df.min())
print("\nStandard Deviation:\n", df.std())
print("\n" + "="*50)
```

```python
# 3. Subject-wise averages + hardest subject

subject_avg = df.mean()
hardest_subject = subject_avg.idxmin()

print(" Subject-wise Averages:\n")
print(subject_avg)
print(f"\n Hardest Subject: {hardest_subject}")
print("\n" + "="*50)

```

```python
# 4. Student-wise averages + struggling students

df["Average"] = df.mean(axis=1)

struggling_students = df[df["Average"] < 75]

print(" Student Averages:\n")
print(df["Average"])

print("\nStruggling Students (Average < 75):\n")
print(struggling_students)
print("\n" + "="*50)

```

```python
# 5. Top 5 performing students

top_5 = df.sort_values(by="Average", ascending=False).head(5)

print("Top 5 Students:\n")
print(top_5)
print("\n" + "="*50)

```

```python
# 6. Add a new student

new_student = pd.DataFrame([[85, 88, 90, 87, 86]],
                           columns=subjects,
                           index=["Student_21"])

df = pd.concat([df, new_student])

# Recalculate average
df["Average"] = df.mean(axis=1)

print(" Added New Student:\n")
print(df.tail())
print("\n" + "="*50)

```

```python
# 7. Remove lowest-performing student

lowest_student = df["Average"].idxmin()
df = df.drop(index=lowest_student)

print(f" Removed Lowest Performing Student: {lowest_student}")
print("\nUpdated Dataset Shape:", df.shape)
print("\n" + "="*50)

```

```python
# 8. Performance Report

print(" FINAL PERFORMANCE REPORT\n")

print(f"Total Students: {len(df)}")
print(f"Hardest Subject: {hardest_subject}")
print(f"Average Class Score: {df['Average'].mean():.2f}")

print("\nTop Performer:")
print(df["Average"].idxmax(), "-", df["Average"].max())

print("\nStudents Needing Attention:")
print(df[df["Average"] < 75].index.tolist())
```
